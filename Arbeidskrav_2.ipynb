### Oppgave 5


```python
"""
Created on Sun Nov  8 12:26:32 2026

@author: Lene Kristin Pedersen
PY1010 Arbeidskrav 2 Oppgave 5)
"""

import numpy as np

a = float(input("Skriv inn diameter for sirkelen:"))
b = float(input("Skriv inn høyden i trekanten:"))

print("Diameter er:", a)
print("Høyde er:", b)

def fun_figur (d, h):    #funksjon
    omkrets_sirkel = 2 * np.pi * d/2   # Omkrets av sirkelen i figuren
    areal_sirkel = np.pi * (d/2)**2    # Areal av sirkelen i figuren
    areal_trekant = d * h / 2   # Arealet av trekanten i figuren
    hyp = np.sqrt(d**2 + h**2)   # Hypotenusen til trekanten
    omkrets = (omkrets_sirkel/2) + h + hyp   # Regner ut omkrets av figuren
    areal = (areal_sirkel/2) + areal_trekant   # Regner ut arealet av figuren
    return (round(omkrets,2), round(areal,2))  # Gir resultat på omkrets og areal med to desimaler

(omkrets, areal) = fun_figur(a, b)   # Utfører funksjonen på innvariablene a (omkrets) og b (areal)

print ("Omkrets av figuren er:", omkrets)
print ("Areal av figuren er:", areal)
```
